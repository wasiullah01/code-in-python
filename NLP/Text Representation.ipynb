Bag of Words

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# Step 1: Create sample texts
texts = [
    "I love Python",
    "I love Machine Learning",
    "Python is great"
]
```

```python
#step 2: Create CountVectorizer
vectorizer = CountVectorizer()
```

```python
#step 3: Fit and Transform
X = vectorizer.fit_transform(texts)
```

```python
#Step 4: Convert to array
X_array = X.toarray()
```

```python
#Step 5: Get Vocabulary
vocab = vectorizer.get_feature_names_out()

print("Vocabulary: ",vocab)
print("Bag of Words Matrix: ")
print(X_array)
```

```python
import pandas as pd

# Texts
texts = [
    "cat sat on mat",
    "cat and dog sat",
    "cat loves the dog"
]

# Create BoW
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

# Convert to DataFrame for clarity
df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())

print("Bag of Words Representation:")
print(df)
```

N gram

N-gram = Sequence of N consecutive words from a text

Why? BoW ignores word order! N-grams capture context & word relationships!

```python
df = pd.DataFrame({
    'texts': [
        'Python is  the great',
        'Machine learning is amazing',
        'Python machine learning'
    ]
})
```

```python
#Uni-gram
cv = CountVectorizer(ngram_range=(1,1))
bigrams = cv.fit_transform(df['texts'])
print(cv.vocabulary_)
print("Unigram: Matrix\n",bigrams.toarray())
print("Unigram Features \n", cv.get_feature_names_out())
```

```python
#Bi-gram
cv = CountVectorizer(ngram_range=(2,2))
bigrams = cv.fit_transform(df['texts'])
print(cv.vocabulary_)
print("Bigram: Matrix\n",bigrams.toarray())
print("Bigram Features \n", cv.get_feature_names_out())
```

```python
#Tri-gram
cv = CountVectorizer(ngram_range=(3,3))
bigrams = cv.fit_transform(df['texts'])
print(cv.vocabulary_)
print("Trigram: Matrix\n",bigrams.toarray())
print("trigram Features \n", cv.get_feature_names_out())
```

```python
#Mixed
cv = CountVectorizer(ngram_range=(1,3))
bigrams = cv.fit_transform(df['texts'])
print(cv.vocabulary_)
print("All N-gram: Matrix\n",bigrams.toarray())
print("All N-gram Features \n", cv.get_feature_names_out())
```

TF-IDF
Term Frequency - Inverse Document Frequency
Give HIGH weight to important words, LOW weight to common words

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
text1 = [
    "I love Python",
    "I love Machine Learning",
    "Python is great"
]

tfidf = TfidfVectorizer()

X = tfidf.fit_transform(text1)

df = pd.DataFrame(X.toarray(), columns=tfidf.get_feature_names_out())

print("TF-IDF Matrix ")
print(df)
print("Vocabulary")
print(tfidf.vocabulary_)
```

Comparing Bag of Words and TF-IDF

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

texts = [
    "Python Python Python",
    "Machine Learning Learning",
    "I I I love love"
]

# BAG OF WORDS
print("BAG OF WORDS:")
bow = CountVectorizer()
X_bow = bow.fit_transform(texts)
df_bow = pd.DataFrame(X_bow.toarray(), columns=bow.get_feature_names_out())
print(df_bow)

# TF-IDF
print("\nTF-IDF:")
tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(texts)
df_tfidf = pd.DataFrame(X_tfidf.toarray(), columns=tfidf.get_feature_names_out())
print(df_tfidf)
```
