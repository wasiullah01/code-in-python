```python
from nltk.tokenize import RegexpTokenizer
from textblob import TextBlob

# Text
text = "Hello! How are you doing today? NLP is amazing!"

# Sentence tokenization without downloading NLTK data
sentences = [sentence.strip() for sentence in text.split("!") if sentence.strip()]
print("Sentences:")
for sent in sentences:
    print(f"  - {sent}!")

# Word tokenization with punctuation as separate tokens
tokenizer = RegexpTokenizer(r"\w+|[^\w\s]")
words = tokenizer.tokenize(text)
print("\nWords:")
print(words)

# TextBlob object creation does not require the missing tokenizer data
blob = TextBlob(text)
print("\nWords (TextBlob-compatible):")
print(words)

print("\nSentences (TextBlob-compatible):")
for sent in sentences:
    print(f"  - {sent}!")
```

Sentiment Analysis

```python

#texts
texts = [
    "I love Python! It's amazing!",
    "This is the worst day ever.",
    "The weather is okay, nothing special.",
    "I hate waiting for the bus."
]

print("SENTIMENT ANALYSIS\n" + "="*50)

for text in texts:
    blob = TextBlob(text)
    polarity = blob.sentiment.polarity  # -1 to 1 (-1=negative, 0=neutral, 1=positive)
    subjectivity = blob.sentiment.subjectivity  # 0=objective, 1=subjective
    
    if polarity > 0.1:
        sentiment = "POSITIVE"
    elif polarity < -0.1:
        sentiment = "NEGATIVE"
    else:
        sentiment = "NEUTRAL"
    
    print(f"Text: {text}")
    print(f"Sentiment: {sentiment}")
    print(f"Polarity: {polarity:.2f}")
    print()
```

```python
from textblob import TextBlob

print("SENTIMENT ANALYZER")
print("="*50)

while True:
    text = input("\nEnter text (or 'quit' to exit): ")
    
    if text.lower() == 'quit':
        print("Goodbye!")
        break
    
    blob = TextBlob(text)
    polarity = blob.sentiment.polarity
    
    if polarity > 0.1:
        sentiment = "POSITIVE "
    elif polarity < -0.1:
        sentiment = "NEGATIVE "
    else:
        sentiment = "NEUTRAL "
    
    print(f"Sentiment: {sentiment}")
    print(f"Polarity: {polarity:.2f}")
```

Word Frequency

```python
from collections import Counter

text = """
Python is great. Python is powerful. Python is easy to learn.
Machine learning is important. Machine learning uses Python.
NLP with Python is amazing. Python NLP is fun!
"""

blob = TextBlob(text.lower())
words = blob.words

# Remove short words
words = [word for word in words if len(word) > 3]

# Count frequency
word_freq = Counter(words)

print("WORD FREQUENCY\n" + "="*50)
for word, count in word_freq.most_common(10):
    print(f"{word}: {count} times")
```
